```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

print("Liberias cargadas correctamente")
```

```python
df = pd.read_csv('/WA_Fn-UseC_-HR-Employee-Attrition.csv')
print(df.shape)
```

```python
df['Attrition'].value_counts()
```

```python
rotacion = df['Attrition'].value_counts(normalize=True) * 100
print(rotacion.round(2))
```

```python
df.groupby('Department')['Attrition'].value_counts()
```

```python
rotacion_dept = df.groupby('Department')['Attrition'].apply(
    lambda x: (x == 'Yes').sum() / len(x) * 100
).round(2)
print(rotacion_dept)
```

```python
plt.figure(figsize=(8, 4))
rotacion_dept.plot(kind='bar', color='steelblue', edgecolor='black')
plt.title('Tasa de rotacion por departamento')
plt.xlabel('Department')
plt.ylabel('% Rotacion')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
salario_Attrition = df.groupby('Attrition')['MonthlyIncome'].mean().round(2)
print(salario_Attrition)
```

```python
plt.figure(figsize=(6, 4))
salario_Attrition.plot(kind='bar', color=['steelblue', 'salmon'], edgecolor='black')
plt.title('Salario promedio por attrition')
plt.xlabel('Attrition')
plt.ylabel('Salario promedio mensual')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
df['Attrition_num'] = (df['Attrition'] == 'Yes').astype(int)
correlacion = df.corr(numeric_only=True)['Attrition_num'].sort_values()
print(correlacion)
```

```python
plt.figure(figsize=(10, 6))
correlacion.drop('Attrition_num').plot(kind='bar', color='steelblue')
plt.title('Correlacion de variables con attrition')
plt.xlabel('Correlacion')
plt.axvline(x=0, color='red', linestyle='--')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(data=df, x='Age', hue='Attrition', bins=20, kde=True)
plt.title('Distribucion de edad por attrition')
plt.xlabel('Edad')
plt.ylabel('Cantidad de empleados')
plt.show()
```

```python
ventas = df[df['Department'] == 'Sales']
print(ventas.shape)
```

```python
sales_que_se_fueron = df[(df['Department'] == 'Sales') & (df['Attrition'] == 'Yes')]
print(sales_que_se_fueron.shape)
```

```python
sales_hr = df[(df['Department'] == 'Sales') | (df['Department'] == 'Human Resources')]
print(sales_hr.shape)
print(sales_hr['Department'].value_counts())
```

```python
import numpy as np
df['Rango_salarial'] = np.where(df['MonthlyIncome'] > 7000, 'Alto',
                               np.where(df['MonthlyIncome'] >= 3000, 'Medio', 'Bajo')
                               )
print(df['Rango_salarial'].value_counts())
```

```python
rotacion_salario = df.groupby('Rango_salarial')['Attrition'].apply(lambda x: (x == 'Yes').sum() / len(x) * 100).round(2)
print(rotacion_salario)
```

```python
plt.figure(figsize = (7, 4))
rotacion_salario.sort_values().plot(kind = 'bar', color =['green', 'orange', 'red'], edgecolor = 'black')
plt.title('Tasa de rotacion por rango salarial')
plt.xlabel('Rango salarial')
plt.ylabel('% rotacion')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# crear un resumen completo
resumen = df.groupby(['Department', 'Rango_salarial']).agg(
          Total_empleados = ('EmployeeNumber', 'count'),
          Salario_promedio = ('MonthlyIncome', 'mean'),
          Tasa_rotacion = ('Attrition_num', 'mean')
).round(2)
resumen['Tasa_rotacion'] = resumen['Tasa_rotacion'] * 100
print(resumen)
```

```python
resumen.to_excel('HR_Resumen_analisis.xlsx')
print("Archivo exportado correctamente")
```
